## Đề bài

Chương trình dưới đây muốn in ``x là 1 hoặc 2'' khi $x\in\{1,2\}$, nhưng với $x=5$ vẫn in như vậy. Giải thích bằng quy tắc ưu tiên toán tử và giá trị ``đúng/sai'' của số, rồi viết *hai* cách sửa đúng.

## Chạy mã ban đầu

````python
x = 5
if x == 1 or 2:
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
````

## Bài làm

Với `x = 5`, chương trình vẫn in `x là 1 hoặc 2`, nên kết quả chưa đúng.

Phép so sánh `==` có độ ưu tiên cao hơn `or`, vì vậy điều kiện được hiểu là:

```python
(x == 1) or 2
```

`x == 1` là `False`, nhưng số 2 có giá trị chân lý là đúng vì khác 0. Cụ thể, `False or 2` trả về 2; khi dùng trong `if`, giá trị này được coi là đúng.

**Cách 1:** Viết đầy đủ hai phép so sánh.

````python
x = 5
if x == 1 or x == 2:
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
````

**Cách 2:** Dùng `in` để kiểm tra `x` có nằm trong `(1, 2)` hay không.

````python
x = 5
if x in (1, 2):
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
````

Kiểm tra cả hai cách với các giá trị thuộc và không thuộc tập cần xét:

````python
for x, du_kien in [(1, True), (2, True), (0, False), (5, False)]:
    cach_1 = x == 1 or x == 2
    cach_2 = x in (1, 2)
    print(f"x = {x}: cách 1 = {cach_1}, cách 2 = {cach_2}")
    assert cach_1 is du_kien
    assert cach_2 is du_kien
````

Cả hai cách đều đúng: điều kiện chỉ đúng khi `x` bằng 1 hoặc 2.
