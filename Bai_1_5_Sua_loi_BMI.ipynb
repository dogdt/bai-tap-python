## Đề bài

Đoạn mã dưới đây tính chỉ số khối cơ thể $\mathrm{BMI}=\text{cân nặng (kg)}/\text{chiều cao (m)}^2$. Chạy với đầu vào $60$ và $1{,}7$, đọc thông báo lỗi, giải thích có *ít nhất hai lỗi khác nhau*, rồi sửa để được $\mathrm{BMI}=20{,}8$.

## Mã ban đầu

```python
can_nang = input("Cân nặng (kg): ")
chieu_cao = input("Chiều cao (m): ")
bmi = can_nang / chieu_cao ^ 2
print(f"BMI = {bmi:.1f}")
```

## Chạy lại lỗi thứ nhất

Để notebook có thể chạy toàn bộ mà không dừng chờ bàn phím, hai biến dưới đây được gán chuỗi `"60"` và `"1.7"`, đúng kiểu dữ liệu mà `input()` trả về. `try/except` chỉ dùng để hiển thị thông báo lỗi và tiếp tục chạy các ô sau.

Khi nhập từ bàn phím Python, chiều cao phải viết `1.7` với dấu chấm.

````python
can_nang = "60"
chieu_cao = "1.7"
try:
    bmi = can_nang / chieu_cao ^ 2
except TypeError as loi:
    print(f"{type(loi).__name__}: {loi}")
````

## Phân tích hai lỗi

1. **Sai kiểu dữ liệu:** `input()` trả về chuỗi. Python không cho phép chia hai chuỗi, nên phép `/` gây `TypeError`. Cần dùng `float()` để chuyển cân nặng và chiều cao sang số.
2. **Sai toán tử:** `^` là phép XOR trên bit, không phải lũy thừa. Bình phương trong Python phải viết `** 2`.

Nếu chỉ sửa lỗi thứ nhất mà vẫn giữ `^`, chương trình vẫn lỗi. Phép `/` có độ ưu tiên cao hơn `^`, nên biểu thức được hiểu thành `(can_nang / chieu_cao) ^ 2`; kết quả bên trái là số thực, không dùng được với phép XOR.

````python
can_nang = float("60")
chieu_cao = float("1.7")
try:
    bmi = can_nang / chieu_cao ^ 2
except TypeError as loi:
    print(f"{type(loi).__name__}: {loi}")
````

## Mã đã sửa

Dùng `float()` để đổi kiểu và `**` để bình phương chiều cao. Dữ liệu kiểm tra vẫn là 60 kg và 1.7 m.

````python
can_nang = float("60")
chieu_cao = float("1.7")
bmi = can_nang / (chieu_cao ** 2)
print(f"BMI = {bmi:.1f}")
````

Nếu chạy tương tác và nhập từ bàn phím, thay hai dòng gán dữ liệu bằng:

```python
can_nang = float(input("Cân nặng (kg): "))
chieu_cao = float(input("Chiều cao (m): "))
```

Chiều cao cần lớn hơn 0. Giá trị đúng là `60 / 1.7**2 ≈ 20.7612456747`, nên in với một chữ số thập phân được `20.8` (viết theo dấu phẩy thập phân là `20,8`).

````python
from math import isclose

assert isclose(bmi, 20.761245674740486, rel_tol=1e-12)
assert f"{bmi:.1f}" == "20.8"
print("Đã kiểm tra: BMI làm tròn một chữ số thập phân bằng 20.8.")
````
