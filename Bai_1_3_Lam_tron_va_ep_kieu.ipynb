## Đề bài

Dự đoán đầu ra của bốn dòng dưới đây rồi chạy để kiểm tra. Chú ý các giá trị làm tròn và phép ép kiểu; giải thích *vì sao* `round(2.5)` và `round(3.5)` không cho kết quả ``tăng một đơn vị'' như nhau.

## Mã của đề bài

```python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)
```

## Dự đoán đầu ra

```text
2 4 0 1.23
-3 -4.0 3
15 123
abababc 7.0 -2
```

- Dòng 1: khi giá trị nằm chính xác ở giữa hai số nguyên, `round()` chọn số nguyên chẵn gần nhất. Vì vậy `round(2.5) = 2`, `round(3.5) = 4` và `round(0.5) = 0`. `round(1.2345, 2)` cho `1.23`.
- Dòng 2: `int()` cắt phần thập phân về phía 0, nên `int(-3.7) = -3`. `-3.7 // 1` lấy phần nguyên dưới, cho `-4.0`; còn `int(3.99) = 3`.
- Dòng 3: sau khi ép kiểu, `12 + 3 = 15`; với chuỗi, dấu `+` nối thành `"123"`.
- Dòng 4: `"ab" * 3 + "c"` cho `"abababc"`; `7 / 2 * 2` tính từ trái sang phải và cho số thực `7.0`; `10 - 2 ** 2 * 3 = 10 - 12 = -2`.

Quy tắc làm tròn về số chẵn áp dụng khi giá trị thực sự nằm đúng giữa hai lựa chọn. Không nên suy rộng rằng mọi số thập phân trong máy đều được biểu diễn chính xác.

## Chạy kiểm tra

````python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)
````

````python
assert (round(2.5), round(3.5), round(0.5)) == (2, 4, 0)
assert (int(-3.7), -3.7 // 1) == (-3, -4.0)
print("Kết quả làm tròn và ép kiểu đúng như dự đoán.")
````

## Nhận xét

`round()`, `int()` và `//` có quy tắc khác nhau. Với số âm, không thể thay `//` bằng `int()` rồi cho rằng kết quả luôn giống nhau.
