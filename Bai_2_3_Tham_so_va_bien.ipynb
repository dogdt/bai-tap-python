## Đề bài

Dự đoán đầu ra, chú ý tham số mặc định và biến cùng tên ở trong/ngoài hàm.

## Mã của đề bài

```python
x = 10
def f(x, y=2):
    x = x + y
    return x * 2

print(f(3), f(3, 5), f(y=1, x=4), x)
```

## Bài làm

Em dự đoán chương trình in ra:

```text
10 16 10 10
```

| Lời gọi | Cách tính | Kết quả |
|---|---|---|
| `f(3)` | `y` mặc định bằng 2, nên `(3 + 2) * 2` | 10 |
| `f(3, 5)` | `(3 + 5) * 2` | 16 |
| `f(y=1, x=4)` | Truyền theo tên, nên `(4 + 1) * 2` | 10 |

Biến `x` trong hàm là biến địa phương. Lệnh `x = x + y` chỉ đổi biến này, nên `x` ở ngoài hàm vẫn bằng 10.

Chạy lại đoạn mã để kiểm tra:

````python
x = 10
def f(x, y=2):
    x = x + y
    return x * 2

print(f(3), f(3, 5), f(y=1, x=4), x)
````

````python
assert f(3) == 10
assert f(3, 5) == 16
assert f(y=1, x=4) == 10
assert x == 10
````

Kết quả chạy đúng với dự đoán. Tham số mặc định được dùng khi lời gọi không truyền giá trị cho `y`.
