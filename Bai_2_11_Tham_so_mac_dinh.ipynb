## Đề bài

Hàm `tao_nhom` cho mỗi lần gọi một nhóm mới chứa tên truyền vào. Dự đoán rồi chạy; vì sao hai nhóm lại giống nhau? Sửa hàm để `a` và `b` là hai danh sách độc lập.

## Bài làm

Em dự đoán mã ban đầu in ra:

```text
['An', 'Bình'] ['An', 'Bình'] True
```

Chạy kiểm tra:

````python
def tao_nhom(ten, thanh_vien=[]):
    thanh_vien.append(ten)
    return thanh_vien

a = tao_nhom("An")
b = tao_nhom("Bình")
print(a, b, a is b)
````

Danh sách mặc định `[]` được tạo một lần khi định nghĩa hàm, không phải tạo lại mỗi lần gọi.

- Lần gọi thứ nhất thêm `"An"`, rồi `a` tham chiếu đến danh sách đó.
- Lần gọi thứ hai vẫn dùng danh sách này và thêm `"Bình"`, rồi `b` cũng tham chiếu đến nó.

Vì vậy, sau hai lần gọi, cả `a` và `b` cùng trỏ tới `['An', 'Bình']`, nên `a is b` là `True`.

**Cách sửa:** Dùng `None` làm giá trị mặc định và tạo danh sách mới bên trong hàm.

````python
def tao_nhom(ten, thanh_vien=None):
    if thanh_vien is None:
        thanh_vien = []
    thanh_vien.append(ten)
    return thanh_vien


a = tao_nhom("An")
b = tao_nhom("Bình")
print(a, b, a is b)
````

````python
assert a == ["An"]
assert b == ["Bình"]
assert a is not b

a.append("Chi")
assert b == ["Bình"]
print("Sau khi thêm Chi vào a:")
print("a =", a)
print("b =", b)
````

Sau khi sửa, hai nhóm được tạo riêng. Thêm thành viên vào `a` không làm thay đổi `b`.
