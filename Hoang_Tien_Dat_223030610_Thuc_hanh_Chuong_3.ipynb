## TH 3.2 — Trung bình cắt đầu–cuối

### Đề bài

Điểm của 8 giám khảo là `[7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]`. Bỏ 2 điểm cao nhất và 2 điểm thấp nhất rồi tính trung bình các điểm còn lại. So sánh với trung bình thông thường và không sửa danh sách gốc.

### Bài làm

Em dùng `sorted()` để tạo danh sách đã sắp xếp, rồi cắt bỏ hai phần tử ở mỗi đầu. Các điểm còn lại là `6.5, 7.0, 7.5, 8.0`, có trung bình 7,25.

```python
diem = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
diem_ban_dau = diem.copy()
n = 2

diem_sap_xep = sorted(diem)
diem_con_lai = diem_sap_xep[n:len(diem_sap_xep) - n]
tb_thuong = sum(diem) / len(diem)
tb_cat = sum(diem_con_lai) / len(diem_con_lai)

print("Danh sách gốc:", diem)
print("Điểm còn lại:", diem_con_lai)
print(f"Trung bình thông thường: {tb_thuong:.4f}")
print(f"Trung bình cắt đầu–cuối: {tb_cat:.4f}")
```

```python
assert diem == diem_ban_dau
assert diem_con_lai == [6.5, 7.0, 7.5, 8.0]
assert tb_thuong == 6.6875
assert tb_cat == 7.25
```

Trung bình thông thường là 6,6875, còn trung bình sau khi cắt là 7,25. Trong bộ dữ liệu này, việc bỏ các điểm cực trị làm trung bình tăng lên; cách cắt đầu–cuối giúp giảm ảnh hưởng của những điểm quá cao hoặc quá thấp.

## TH 3.4 — Hồi quy tuyến tính

### Đề bài

Số giờ ôn tập và điểm thi của 5 sinh viên là `[(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]`. Tìm đường hồi quy `y = a + bx` theo công thức:

$$b=\frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{\sum (x_i-\bar{x})^2},\qquad a=\bar{y}-b\bar{x}.$$

Dự đoán điểm khi ôn 6 giờ và kiểm tra bằng `statistics.linear_regression`.

### Bài làm

Em tính trung bình của x và y trước, rồi tính tử số và mẫu số của hệ số b. Với dữ liệu đề cho, hai tổng này đều bằng 10 nên b = 1.

```python
import statistics
from math import isclose

du_lieu = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]
ds_x = [cap[0] for cap in du_lieu]
ds_y = [cap[1] for cap in du_lieu]
x_tb = sum(ds_x) / len(ds_x)
y_tb = sum(ds_y) / len(ds_y)

tu_so = sum((x - x_tb) * (y - y_tb) for x, y in du_lieu)
mau_so = sum((x - x_tb)**2 for x in ds_x)
b = tu_so / mau_so
a = y_tb - b * x_tb
du_doan = a + b * 6

kiem_tra = statistics.linear_regression(ds_x, ds_y)
print(f"x trung bình = {x_tb:.1f}; y trung bình = {y_tb:.1f}")
print(f"Tử số = {tu_so:.1f}; mẫu số = {mau_so:.1f}")
print(f"a = {a:.1f}; b = {b:.1f}")
print(f"Đường hồi quy: y = {a:.1f} + {b:.1f}x")
print(f"Dự đoán khi ôn 6 giờ: {du_doan:.1f} điểm")
print(f"Kiểm tra bằng statistics: a = {kiem_tra.intercept:.1f}; b = {kiem_tra.slope:.1f}")
```

```python
assert isclose(a, 3.2, abs_tol=1e-12)
assert isclose(b, 1.0, abs_tol=1e-12)
assert isclose(du_doan, 9.2, abs_tol=1e-12)
assert isclose(a, kiem_tra.intercept, abs_tol=1e-12)
assert isclose(b, kiem_tra.slope, abs_tol=1e-12)
```

Kết quả là `y = 3.2 + x`; dự đoán khi ôn 6 giờ được 9,2 điểm. Vì 6 giờ nằm ngoài khoảng dữ liệu từ 1 đến 5 giờ, đây là dự đoán ngoại suy nên chỉ có tính tham khảo.

## TH 3.6 — Gom sinh viên theo điểm chữ

### Đề bài

Cho các cặp tên và điểm: `[("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]`.

Quy đổi điểm: từ 8,5 là A; từ 7 là B; từ 5,5 là C; từ 4 là D; còn lại F. Gom tên theo điểm chữ bằng `defaultdict(list)` và in các nhóm theo thứ tự chữ cái.

### Bài làm

Em xét các ngưỡng từ cao xuống thấp, rồi thêm tên vào danh sách tương ứng với điểm chữ. Dùng `sorted(nhom)` để in các nhóm theo thứ tự A, B, C, D, F.

```python
from collections import defaultdict


def diem_chu(diem):
    if diem >= 8.5:
        return "A"
    elif diem >= 7:
        return "B"
    elif diem >= 5.5:
        return "C"
    elif diem >= 4:
        return "D"
    return "F"


sinh_vien = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6),
             ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]
nhom = defaultdict(list)
for ten, diem in sinh_vien:
    nhom[diem_chu(diem)].append(ten)

for chu in sorted(nhom):
    print(f"{chu}: {', '.join(nhom[chu])}")
```

```python
assert dict(nhom) == {
    "A": ["An", "Chi"],
    "B": ["Binh", "Giang"],
    "C": ["Em"],
    "D": ["Dung"],
    "F": ["Phuc"]
}
assert [diem_chu(d) for d in [8.5, 7.0, 5.5, 4.0, 3.9]] == ["A", "B", "C", "D", "F"]
```

Nhóm A có An và Chi; nhóm B có Binh và Giang; các nhóm C, D, F lần lượt có Em, Dung, Phuc. Các trường hợp đúng tại ngưỡng cũng được xếp vào nhóm tương ứng.

## TH 3.8 — Hợp nhất doanh thu hai cửa hàng

### Đề bài

Cửa hàng 1 có doanh thu `"ao": 120, "quan": 80, "non": 30`; cửa hàng 2 có `"ao": 90, "giay": 150, "non": 20`. Hợp nhất thành tổng doanh thu theo sản phẩm và in giảm dần. Dùng phép toán tập hợp trên khóa để tìm sản phẩm bán ở cả hai cửa hàng và sản phẩm chỉ bán ở một cửa hàng.

### Bài làm

Em sao chép dữ liệu cửa hàng 1 rồi cộng thêm doanh thu cửa hàng 2 bằng `get()`. Giao của hai tập khóa là sản phẩm bán ở cả hai; hiệu đối xứng là sản phẩm chỉ bán ở một cửa hàng.

```python
cua_hang_1 = {"ao": 120, "quan": 80, "non": 30}
cua_hang_2 = {"ao": 90, "giay": 150, "non": 20}

tong_doanh_thu = cua_hang_1.copy()
for san_pham, doanh_thu in cua_hang_2.items():
    tong_doanh_thu[san_pham] = tong_doanh_thu.get(san_pham, 0) + doanh_thu

sap_xep = sorted(tong_doanh_thu.items(), key=lambda cap: cap[1], reverse=True)
print("Tổng doanh thu theo thứ tự giảm dần:")
for san_pham, doanh_thu in sap_xep:
    print(f"{san_pham}: {doanh_thu}")

ban_ca_hai = cua_hang_1.keys() & cua_hang_2.keys()
chi_mot_noi = cua_hang_1.keys() ^ cua_hang_2.keys()
print("Bán ở cả hai cửa hàng:", sorted(ban_ca_hai))
print("Chỉ bán ở một cửa hàng:", sorted(chi_mot_noi))
```

```python
assert tong_doanh_thu == {"ao": 210, "quan": 80, "non": 50, "giay": 150}
assert sap_xep == [("ao", 210), ("giay", 150), ("quan", 80), ("non", 50)]
assert ban_ca_hai == {"ao", "non"}
assert chi_mot_noi == {"quan", "giay"}
assert sum(tong_doanh_thu.values()) == sum(cua_hang_1.values()) + sum(cua_hang_2.values())
assert cua_hang_1 == {"ao": 120, "quan": 80, "non": 30}
```

Áo có tổng doanh thu cao nhất là 210. Áo và nón bán ở cả hai cửa hàng; quần và giày chỉ xuất hiện ở một cửa hàng. Doanh thu gốc của cửa hàng 1 vẫn được giữ nguyên.

## TH 3.10 — Sàng Eratosthenes bằng set

### Đề bài

Dùng sàng Eratosthenes, cài đặt bằng `set`, tìm các số nguyên tố không vượt quá 1000. In số lượng, tổng và số cặp số nguyên tố sinh đôi `(p, p + 2)` cùng không vượt quá 1000. Kiểm tra: có 168 số nguyên tố, tổng bằng 76 127.

### Bài làm

Ban đầu em đưa các số từ 2 đến 1000 vào một tập hợp. Với mỗi số nguyên tố p không vượt quá căn bậc hai của 1000, loại các bội từ `p*p`. Những bội nhỏ hơn `p*p` đã được loại bởi các số nguyên tố nhỏ hơn.

Sau khi sàng, em xét mỗi p trong tập; nếu `p + 2` cũng thuộc tập thì có một cặp sinh đôi.

```python
from math import isqrt

N = 1000
nguyen_to = set(range(2, N + 1))

for p in range(2, isqrt(N) + 1):
    if p in nguyen_to:
        nguyen_to.difference_update(range(p * p, N + 1, p))

ds_nguyen_to = sorted(nguyen_to)
cap_sinh_doi = [(p, p + 2) for p in ds_nguyen_to if p + 2 in nguyen_to]
print("Số lượng số nguyên tố:", len(nguyen_to))
print("Tổng các số nguyên tố:", sum(nguyen_to))
print("Số cặp nguyên tố sinh đôi:", len(cap_sinh_doi))
```

```python
assert len(nguyen_to) == 168
assert sum(nguyen_to) == 76_127
assert len(cap_sinh_doi) == 35
assert cap_sinh_doi[0] == (3, 5)
assert all(p + 2 <= N for p, q in cap_sinh_doi)
```

Có 168 số nguyên tố không vượt quá 1000, tổng 76 127 và 35 cặp nguyên tố sinh đôi. Mỗi cặp được đếm một lần theo số nhỏ hơn của cặp.
