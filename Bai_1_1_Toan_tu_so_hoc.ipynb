## Đề bài

Với `x = 23` và `y = 6`, hãy tính tay giá trị các biểu thức `x / y`, `x // y`, `x % y`, `x ** 2`,
`-x // y`, `-x % y`, `2 ** 3 ** 2`, `(2 ** 3) ** 2`; sau đó kiểm tra lại bằng máy.

## Dự đoán và giải thích

Với `x = 23`, `y = 6`:

| Biểu thức | Kết quả dự đoán | Giải thích |
|---|---|---|
| `x / y` | `3.8333333333333335` | Phép chia thông thường, kết quả là số thực. |
| `x // y` | `3` | Lấy phần nguyên dưới của thương. |
| `x % y` | `5` | Vì `23 = 6 * 3 + 5`. |
| `x ** 2` | `529` | Bình phương của 23. |
| `-x // y` | `-4` | Làm tròn xuống của `-23/6`, về phía âm vô cùng. |
| `-x % y` | `1` | Vì `-23 = 6 * (-4) + 1`. |
| `2 ** 3 ** 2` | `512` | Lũy thừa kết hợp từ phải: `2 ** (3 ** 2)`. |
| `(2 ** 3) ** 2` | `64` | Dấu ngoặc buộc tính `2 ** 3` trước. |

Điểm dễ nhầm: `//` với số âm khác với việc cắt bỏ phần thập phân.

## Chạy kiểm tra

```python
x, y = 23, 6
print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** 2 =", x ** 2)
print("-x // y =", -x // y)
print("-x % y =", -x % y)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```

```python
assert x == y * (x // y) + x % y
assert -x == y * (-x // y) + (-x % y)
assert (-x // y, -x % y) == (-4, 1)
assert 2 ** 3 ** 2 == 512
assert (2 ** 3) ** 2 == 64
print("Các kết quả kiểm tra đều đúng.")
```

## Nhận xét

Kết quả chạy khớp với dự đoán. Phép chia lấy dư phải đi cùng phép chia lấy nguyên để thỏa mãn `a = b * (a // b) + a % b` khi `b != 0`.
