## Đề bài

Dự đoán chính xác những gì chương trình dưới đây in ra (chú ý khoảng trắng, dấu xuống dòng và định dạng số), rồi chạy kiểm tra.

## Mã của đề bài

```python
a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
```

## Dự đoán đầu ra

```text
72
3.5 | 3
7+2=9; 3.5
ab ababab
1,234.57 005 25.6%
```

1. `sep=""` bỏ ký tự ngăn cách giữa 7 và 2, nên in `72`.
2. `sep=" | "` đặt một dấu cách ở mỗi bên dấu `|`.
3. `end="; "` thay ký tự xuống dòng bằng dấu chấm phẩy và một dấu cách. Lệnh `print()` tiếp theo tiếp tục trên cùng dòng; `:.1f` in một chữ số sau dấu thập phân.
4. Hai chuỗi viết liền `"a" "b"` được ghép thành `"ab"`; `3 * "ab"` cho `"ababab"`. `print()` mặc định chèn một dấu cách giữa hai đối số.
5. `:,.2f` có dấu phẩy phân nhóm hàng nghìn và hai chữ số thập phân; `:03d` đệm số 0 bên trái để đủ ba chữ số; `:.1%` đổi sang phần trăm và giữ một chữ số thập phân.

Chương trình có sáu lệnh `print()` nhưng chỉ tạo năm dòng đầu ra vì hai lệnh giữa dùng chung một dòng.

## Chạy kiểm tra

````python
a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
````

## Nhận xét

Đầu ra khớp cả nội dung, dấu cách và số dòng. Dấu `+` nằm trong phần văn bản của f-string được in nguyên dạng; biểu thức trong `{a + b}` mới được tính.
