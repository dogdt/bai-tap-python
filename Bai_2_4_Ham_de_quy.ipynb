## Đề bài

Hàm đệ quy dưới đây trả về cặp `(giá trị, số lời gọi)`. Dự đoán kết quả của `g(2468)`, `g(0)` và `g(1999)`. Hàm `g` thực chất tính đại lượng gì của $n$?

## Mã của đề bài

```python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
```

## Bài làm

Em dự đoán đầu ra là:

```text
(20, 5) (0, 1) (28, 5)
```

Hàm `g` tính tổng các chữ số của số nguyên không âm `n`, đồng thời đếm số lần gọi hàm, kể cả lần gọi đầu tiên và lần gọi với `n = 0`.

Với `g(2468)`, các lời gọi lần lượt là:

```text
g(2468) → g(246) → g(24) → g(2) → g(0)
```

Khi quay lui, kết quả được tính như sau:

| Lời gọi | Tổng chữ số | Số lời gọi |
|---|---:|---:|
| `g(0)` | 0 | 1 |
| `g(2)` | 2 | 2 |
| `g(24)` | 2 + 4 = 6 | 3 |
| `g(246)` | 6 + 6 = 12 | 4 |
| `g(2468)` | 12 + 8 = 20 | 5 |

`g(0)` trả ngay `(0, 1)`. Với `g(1999)`, tổng chữ số là `1 + 9 + 9 + 9 = 28`, có 5 lời gọi tương tự.

Chạy kiểm tra:

````python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
````

````python
assert g(2468) == (20, 5)
assert g(0) == (0, 1)
assert g(1999) == (28, 5)
assert g(7) == (7, 2)
assert g(1000) == (1, 5)
````

Với `n > 0` có `k` chữ số, hàm có `k + 1` lời gọi. `n // 10` bỏ chữ số cuối, còn `n % 10` lấy chữ số cuối để cộng vào tổng.
