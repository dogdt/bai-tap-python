## Đề bài

Chương trình tính $100/\text{số_bạn}$ phải xử lý được cả đầu vào không phải số (`abc`) lẫn $0$. Chạy mã với hai đầu vào đó, giải thích các lỗi còn sót (có *hai* lỗi khác nhau) và sửa.

## Mã ban đầu

```python
try:
    so_ban = int(input("Số bạn: "))
    tb = 100 / so_ban
except ValueError:
    print("Hãy nhập số nguyên!")
print(f"Kết quả: {tb}")
```

## Bài làm

Em kiểm tra với hai chuỗi đầu vào `"abc"` và `"0"`. Các ô dưới gán sẵn chuỗi thay cho `input()`; khối `try/except` ngoài cùng chỉ để in lỗi và cho notebook chạy tiếp.

**Trường hợp nhập `abc`:**

````python
du_lieu = "abc"
try:
    try:
        so_ban = int(du_lieu)
        tb = 100 / so_ban
    except ValueError:
        print("Hãy nhập số nguyên!")
    print(f"Kết quả: {tb}")
except (NameError, ZeroDivisionError) as loi:
    print(f"{type(loi).__name__}: {loi}")
````

`int("abc")` gây `ValueError`, nên chương trình in `Hãy nhập số nguyên!`. Tuy nhiên, dòng in kết quả vẫn chạy dù `tb` chưa được gán, làm phát sinh `NameError`.

**Trường hợp nhập `0`:**

````python
du_lieu = "0"
try:
    try:
        so_ban = int(du_lieu)
        tb = 100 / so_ban
    except ValueError:
        print("Hãy nhập số nguyên!")
    print(f"Kết quả: {tb}")
except (NameError, ZeroDivisionError) as loi:
    print(f"{type(loi).__name__}: {loi}")
````

Phép chia `100 / 0` gây `ZeroDivisionError`. Mã ban đầu chỉ bắt `ValueError`, nên chưa xử lý được lỗi chia cho 0.

Như vậy có hai chỗ cần sửa: bắt thêm `ZeroDivisionError` và chuyển dòng in kết quả vào `else` để chỉ in khi tính toán thành công.

Em viết thành hàm để thử được nhiều đầu vào:

````python
def tinh_tb(du_lieu):
    try:
        so_ban = int(du_lieu)
        tb = 100 / so_ban
    except ValueError:
        print("Hãy nhập số nguyên!")
    except ZeroDivisionError:
        print("Số bạn phải khác 0!")
    else:
        print(f"Kết quả: {tb}")


for du_lieu in ["abc", "0", "4"]:
    print(f"Số bạn: {du_lieu}")
    tinh_tb(du_lieu)
````

Sau khi sửa, nhập `abc` hoặc `0` đều nhận được thông báo phù hợp, còn nhập `4` cho kết quả `25.0`.

Khi muốn nhập trực tiếp từ bàn phím, gọi:

```python
tinh_tb(input("Số bạn: "))
```

Khi kiểm tra mã cũ trong notebook, cần chạy từ đầu với kernel mới. Nếu `tb` còn từ lần chạy trước, dòng in có thể dùng nhầm kết quả cũ và che mất lỗi chưa gán biến.
