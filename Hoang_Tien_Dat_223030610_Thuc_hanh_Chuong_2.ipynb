## TH 2.3

### Đề bài

Viết chương trình tính lương tuần như Ví dụ 2.1 (lương 100 000 đồng/giờ, giờ vượt 40 tính gấp 1,5) nhưng nhập lại cho đến khi người dùng gõ một số không âm: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ `abc`, `-2`, `45`.

### Bài làm

Em dùng `while True` để nhập lại. Nếu nhập chữ thì bắt `ValueError`; nếu nhập số âm thì báo lỗi và tiếp tục vòng lặp. Khi hợp lệ mới tính lương.

Với 45 giờ, lương dự kiến là `40 * 100000 + 5 * 150000 = 4 750 000` đồng.

```python
def tinh_luong(gio):
    if gio <= 40:
        return gio * 100_000
    return 40 * 100_000 + (gio - 40) * 150_000


while True:
    try:
        gio = float(input("Nhập số giờ làm trong tuần: "))
    except ValueError:
        print("Lỗi: hãy nhập một số!")
        continue
    if gio < 0:
        print("Lỗi: số giờ phải không âm!")
        continue
    break

luong = tinh_luong(gio)
print(f"Lương tuần: {luong:,.0f} đồng")
```

```python
assert gio == 45
assert luong == 4_750_000
assert tinh_luong(0) == 0
assert tinh_luong(40) == 4_000_000
assert tinh_luong(41) == 4_150_000
```

Chương trình hỏi lại sau cả hai đầu vào `abc` và `-2`; đến `45` thì chấp nhận và in đúng lương 4 750 000 đồng.

## TH 2.4

### Đề bài

Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ `xong`. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo “chưa có dữ liệu”. Kiểm tra với dãy gõ: `7.5`, `8`, `abc`, `6.5`, `9`, `xong`.

### Bài làm

Mỗi số hợp lệ được cộng vào tổng và cập nhật số lớn nhất, nhỏ nhất. Giá trị `abc` bị bỏ qua. Em kiểm tra số lượng trước khi tính trung bình để tránh chia cho 0.

Dữ liệu mẫu có 4 số hợp lệ, tổng 31, trung bình 7,75, lớn nhất 9 và nhỏ nhất 6,5.

```python
def thong_ke_nhap():
    dem = 0
    tong = 0.0
    lon_nhat = None
    nho_nhat = None

    while True:
        chuoi = input("Nhập số thực (gõ xong để dừng): ").strip()
        if chuoi.lower() == "xong":
            break
        try:
            so = float(chuoi)
        except ValueError:
            print("Giá trị không phải số, bỏ qua!")
            continue

        dem += 1
        tong += so
        if lon_nhat is None or so > lon_nhat:
            lon_nhat = so
        if nho_nhat is None or so < nho_nhat:
            nho_nhat = so

    if dem == 0:
        print("chưa có dữ liệu")
        return None

    trung_binh = tong / dem
    print("Số giá trị hợp lệ:", dem)
    print("Tổng:", tong)
    print("Trung bình:", trung_binh)
    print("Lớn nhất:", lon_nhat)
    print("Nhỏ nhất:", nho_nhat)
    return dem, tong, trung_binh, lon_nhat, nho_nhat


ket_qua = thong_ke_nhap()
assert ket_qua == (4, 31.0, 7.75, 9.0, 6.5)
```

Thử thêm trường hợp gõ `xong` ngay từ đầu:

```python
ket_qua_rong = thong_ke_nhap()
assert ket_qua_rong is None
```

Kết quả khớp tính tay. Khi chưa có số hợp lệ, chương trình báo “chưa có dữ liệu” và không thực hiện phép chia.

## TH 2.7

### Đề bài

Dãy Collatz: từ n, nếu n chẵn thì n ← n/2, nếu lẻ thì n ← 3n + 1, lặp đến khi n = 1. Với n = 27, đếm số bước và tìm giá trị lớn nhất mà dãy đạt được (đáp án biết trước: 111 bước, giá trị lớn nhất 9232).

### Bài làm

Em dùng vòng `while` đến khi `n = 1`. Mỗi lần cập nhật `n` được tính là một bước; giá trị lớn nhất được cập nhật sau mỗi bước. Dùng `//` khi chia 2 để giữ `n` là số nguyên.

```python
n = 27
so_buoc = 0
lon_nhat = n

while n != 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    so_buoc += 1
    lon_nhat = max(lon_nhat, n)

print("Số bước:", so_buoc)
print("Giá trị lớn nhất:", lon_nhat)
print("Giá trị cuối:", n)

assert so_buoc == 111
assert lon_nhat == 9232
assert n == 1
```

Từ 27, dãy đạt đến 1 sau 111 bước và đạt giá trị lớn nhất 9232, đúng với đáp án đề cho.

## TH 2.10

### Đề bài

Mở rộng Ví dụ 2.6: ước lượng π với N = 10², 10⁴, 10⁶ điểm (dùng `random.seed(2026)` một lần ở đầu). In ước lượng và sai số tuyệt đối; kiểm tra bằng `assert` rằng sai số nhỏ hơn 6/√N. Nhận xét: gấp 100 lần số điểm thì sai số giảm khoảng mấy lần?

### Bài làm

Em sinh các điểm trong hình vuông `[-1, 1] × [-1, 1]`. Điểm nằm trong hình tròn đơn vị nếu `x*x + y*y <= 1`. Tỉ lệ diện tích hình tròn và hình vuông là π/4, nên ước lượng π bằng `4 * trong / N`.

Đặt seed một lần trước vòng lặp; với mỗi N, bộ đếm điểm trong hình tròn bắt đầu lại từ 0.

```python
import math
import random

random.seed(2026)
print(f"{'N':>10} {'Ước lượng pi':>16} {'Sai số tuyệt đối':>18} {'6/sqrt(N)':>14}")

for N in [10**2, 10**4, 10**6]:
    trong = 0
    for _ in range(N):
        x, y = random.uniform(-1, 1), random.uniform(-1, 1)
        if x * x + y * y <= 1:
            trong += 1

    pi_uoc_luong = 4 * trong / N
    sai_so = abs(pi_uoc_luong - math.pi)
    nguong = 6 / math.sqrt(N)
    print(f"{N:>10,d} {pi_uoc_luong:>16.8f} {sai_so:>18.8f} {nguong:>14.8f}")
    assert sai_so < nguong
```

Sai số điển hình của Monte Carlo có bậc `1/sqrt(N)`. Vì vậy, khi tăng số điểm 100 lần, sai số thường giảm khoảng 10 lần.

Đây là xu hướng thống kê, không phải mỗi lần chạy đều giảm đúng 10 lần hoặc luôn giảm đều. Cả ba trường hợp ở trên đều thỏa ngưỡng kiểm tra của đề.

## TH 2.11

### Đề bài

Phương trình f(x) = x³ − x − 2 = 0 có đúng một nghiệm trong [1, 2] vì f(1) = −2 < 0 < 4 = f(2) và f tăng trên đoạn này. Cài đặt phương pháp chia đôi: giữ đoạn [a, b] mà f(a)f(b) < 0, thay bằng nửa đoạn chứa nghiệm cho tới khi b − a < 10⁻¹⁰. In nghiệm, số bước, và so sánh với kết quả lý thuyết: số bước cần thiết là ceil(log₂(1/10⁻¹⁰)).

### Bài làm

Ở mỗi bước, em lấy trung điểm `m`. Nếu `f(a)` và `f(m)` trái dấu thì giữ `[a, m]`; ngược lại giữ `[m, b]`. Sau khi dừng, lấy trung điểm đoạn cuối làm nghiệm gần đúng.

Độ dài đoạn ban đầu bằng 1. Sau k lần chia đôi, độ dài còn `1/2**k`. Do đó cần 34 bước để nhỏ hơn `10**(-10)`.

```python
import math


def f(x):
    return x**3 - x - 2


a, b = 1.0, 2.0
sai_so = 1e-10
so_buoc = 0

while b - a >= sai_so:
    m = (a + b) / 2
    so_buoc += 1
    if f(m) == 0:
        a = b = m
        break
    if f(a) * f(m) < 0:
        b = m
    else:
        a = m

nghiem = (a + b) / 2
so_buoc_ly_thuyet = math.ceil(math.log2(1 / sai_so))
print(f"Nghiệm gần đúng: {nghiem:.12f}")
print("Số bước thực tế:", so_buoc)
print("Số bước lý thuyết:", so_buoc_ly_thuyet)
print(f"Độ dài đoạn cuối: {b - a:.12e}")
print(f"f(nghiệm) = {f(nghiem):.12e}")

assert b - a < sai_so
assert so_buoc == so_buoc_ly_thuyet == 34
assert f(a) * f(b) < 0
assert abs(nghiem - 1.5213797068) < 1e-10
```

Nghiệm gần đúng là 1,5213797068. Thuật toán dùng 34 bước, khớp với số bước tính theo lý thuyết; sai số của trung điểm không vượt quá nửa độ dài đoạn cuối.
